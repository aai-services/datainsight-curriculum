```python
import sys
import pandas as pd
import matplotlib
import matplotlib.pyplot as plt

print('Python', sys.version.split()[0])
print('pandas', pd.__version__)
print('matplotlib', matplotlib.__version__)
```

```python
rainfall = pd.DataFrame({
    'month': ['Jan', 'Feb', 'Mar', 'Apr'],
    'rainfall_mm': [12, 30, 85, 140],
})
rainfall
```

```python
ax = rainfall.plot(x='month', y='rainfall_mm', kind='bar', legend=False)
ax.set_title('Example: monthly rainfall')
ax.set_xlabel('Month')
ax.set_ylabel('Rainfall (mm)')
plt.show()
```

```python
assert rainfall['rainfall_mm'].sum() == 267
print('Environment OK')
```
