## Exercise 6: fix the bug

This function should add up rainfall values and **skip** missing values (`None`), but it has two bugs. Find and fix them. Before changing anything, predict what `total_rainfall([1, None, 3])` does now.

```python
def total_rainfall(values):
    """Return the total of the values, ignoring any that are None."""
    total = 0
    for v in values:
        if v == None:
            total = 0
        total = total + v
    return total
```

<summary>Hint 1</summary>

```python
# Check yourself
check(lambda: total_rainfall([1, None, 3]) == 4,
      'total_rainfall([1, None, 3]) should be 4.')
```
