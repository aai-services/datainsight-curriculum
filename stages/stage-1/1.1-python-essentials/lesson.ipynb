## Values and variables

A **variable** is a name for a value. Python has a few basic types of value:

| Type | Example | Used for |
|---|---|---|
| `int` | `42` | Whole numbers |
| `float` | `36.6` | Numbers with decimals |
| `str` | `"Riverside"` | Text |
| `bool` | `True` | Yes or no |

```python
# Illustrative numbers for an imaginary town
city = "Riverside"
population = 84_500           # underscores make large numbers readable; Python ignores them
area_km2 = 31.4
is_capital = True

print(city, population, area_km2, is_capital)
print(type(population), type(area_km2))
```

Arithmetic works as you would expect. Note that `/` always gives a `float`, and `**` is a power.

```python
density = population / area_km2
print(density)
print(round(density, 1))
print(2 ** 10)
```

**f-strings** put values into text. Put an `f` before the quotes and the variable in braces. `:,.0f` means "use a thousands separator and no decimals".

```python
print(f"{city} has about {density:,.0f} people per square kilometre.")
```

## Lists

A **list** holds values in order. Positions start at **0**, not 1.

```python
rainfall_mm = [12, 30, 85, 140, 160, 90]     # illustrative values, January to June

print(rainfall_mm[0])       # first item
print(rainfall_mm[-1])      # last item
print(rainfall_mm[1:3])     # items at positions 1 and 2 (the end position is not included)
print(len(rainfall_mm))     # how many items

rainfall_mm.append(40)      # add July
print(rainfall_mm)
```

## Dictionaries

A **dictionary** maps keys to values, like a lookup table.

```python
capitals = {"Ghana": "Accra", "Kenya": "Nairobi", "Nepal": "Kathmandu"}

print(capitals["Kenya"])
capitals["Peru"] = "Lima"                   # add an entry
print(capitals.get("Chile", "not listed"))  # .get gives a default instead of an error
```

## Decisions with `if`

Indentation (four spaces) marks which lines belong to each branch.

```python
month_rain = 85

if month_rain < 10:
    label = "dry"
elif month_rain < 100:
    label = "moderate"
else:
    label = "wet"

print(label)
```

## Repeating with `for`

A `for` loop runs the indented block once for each item.

```python
total = 0
for mm in rainfall_mm:
    total = total + mm

print(f"Total rainfall: {total} mm")
```

## Functions

A **function** packages steps under a name so you can reuse them. `return` sends the result back. The text in triple quotes (a *docstring*) says what the function does.

```python
def mean(values):
    """Return the average of a list of numbers."""
    return sum(values) / len(values)


print(mean(rainfall_mm))
print(mean([1, 2, 3, 4]))
```

## Worked example: a monthly rainfall summary

We combine everything above to summarise rainfall: the average, the wettest month, and how many months were wet.

```python
months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun", "Jul"]


def summarise_rainfall(months, rainfall, wet_threshold=100):
    """Print a short summary of monthly rainfall."""
    wettest_month = months[0]
    wettest_amount = rainfall[0]
    wet_months = 0
    for month, mm in zip(months, rainfall):     # zip walks through two lists together
        if mm > wettest_amount:
            wettest_month, wettest_amount = month, mm
        if mm > wet_threshold:
            wet_months += 1
    print(f"Average: {mean(rainfall):.1f} mm per month")
    print(f"Wettest: {wettest_month} ({wettest_amount} mm)")
    print(f"Months above {wet_threshold} mm: {wet_months}")


summarise_rainfall(months, rainfall_mm)
```

**Check the result.** Before trusting code, check it against something you can work out another way. Here Python's built-in `max` should agree with our loop.

```python
assert max(rainfall_mm) == 160
print("The wettest amount matches max().")
```

## Reading error messages

Errors are normal. When Python stops with an error, the last line of the message says **what** went wrong, and the lines above it show **where**.

The cell below uses `try` and `except` to catch an error so the notebook keeps running; you will not need this yourself yet. To see a real error message, copy `mean([])` into a new cell and run it.

```python
try:
    print(mean([]))
except ZeroDivisionError as err:
    print("ZeroDivisionError:", err)
    print("An empty list has length 0, and Python cannot divide by zero.")
```

Common errors you will meet this week:

| Error | Usual cause |
|---|---|
| `NameError` | A variable used before it was created, or a typo in its name |
| `TypeError` | Mixing types, such as adding a number to text |
| `IndexError` | Asking for a list position that does not exist |
| `KeyError` | Asking a dictionary for a key it does not have |
| `IndentationError` | Missing or inconsistent spaces at the start of a line |
